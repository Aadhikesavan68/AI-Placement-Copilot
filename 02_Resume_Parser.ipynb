```python
import os
import re
import pandas as pd
```

```python
BASE_PATH = "/content/drive/MyDrive/AI-Placement-Copilot-ML"

job_path = os.path.join(BASE_PATH, "datasets/job_descriptions/postings.csv")
processed_path = os.path.join(BASE_PATH, "processed")
```

```python
df_jobs = pd.read_csv(job_path)

print("Total Jobs:", len(df_jobs))
df_jobs.head()
```

```python
df_jobs.info()
```

```python
df_jobs = df_jobs[
    [
        "job_title",
        "company",
        "job_location",
        "job_link",
        "first_seen",
        "search_city",
        "search_country",
        "job level",
        "job_type",
        "job_summary",
        "job_skills"
    ]
]

df_jobs.head()
```

```python
df_jobs.isnull().sum()
```

```python
df_jobs["job_summary"] = df_jobs["job_summary"].fillna("")
df_jobs["job_skills"] = df_jobs["job_skills"].fillna("")
```

```python
def clean_text(text):
    text = str(text)

    text = re.sub(r"\n+", " ", text)
    text = re.sub(r"\t+", " ", text)
    text = re.sub(r" +", " ", text)
    text = re.sub(r"http\S+", "", text)
    text = re.sub(r"www\S+", "", text)

    return text.strip()
```

```python
df_jobs["job_summary"] = df_jobs["job_summary"].apply(clean_text)
df_jobs["job_skills"] = df_jobs["job_skills"].apply(clean_text)

df_jobs.head()
```

```python
df_jobs.info()
df_jobs.head()
```

```python
save_path = os.path.join(processed_path, "clean_jobs.csv")

df_jobs.to_csv(save_path, index=False)

print("Clean job dataset saved successfully!")
```
