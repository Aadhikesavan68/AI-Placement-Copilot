```python
import os
import glob
import pandas as pd
```

```python
BASE = "/content/drive/MyDrive/AI-Placement-Copilot-ML/datasets/processed"

company_root = os.path.join(
    BASE,
    "leetcode-company-wise-interview-questions-main",
    "leetcode-company-wise-interview-questions-main"
)
```

```python
company_folders = [
    folder
    for folder in os.listdir(company_root)
    if os.path.isdir(os.path.join(company_root, folder))
]

print("Companies:", len(company_folders))
print(company_folders[:10])
```

```python
dfs = []

for company in company_folders:

    csv_path = os.path.join(
        company_root,
        company,
        "all.csv"
    )

    if os.path.exists(csv_path):

        df = pd.read_csv(csv_path)

        df["company"] = company

        dfs.append(df)

print("Loaded:", len(dfs))
```

```python
company_df = pd.concat(dfs, ignore_index=True)

print(company_df.shape)

company_df.head()
```

```python
company_df.columns = [
    col.strip().lower().replace(" ", "_")
    for col in company_df.columns
]

company_df.drop_duplicates(inplace=True)

company_df.reset_index(drop=True, inplace=True)
```

```python
company_df.to_csv(
    "/content/drive/MyDrive/AI-Placement-Copilot-ML/datasets/processed/company_questions.csv",
    index=False
)

print("Company dataset saved!")
```
