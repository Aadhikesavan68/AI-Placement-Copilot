```python
import os
import ast
import pandas as pd
```

```python
BASE_PATH = "/content/drive/MyDrive/AI-Placement-Copilot-ML"

matched_jobs_path = os.path.join(
    BASE_PATH,
    "outputs",
    "matched_jobs.csv"
)

output_path = os.path.join(
    BASE_PATH,
    "outputs"
)
```

```python
df = pd.read_csv(matched_jobs_path)

print("Total Records:", len(df))
df.head()
```

```python
def convert_resume_skills(text):

    if pd.isna(text):
        return []

    try:
        return ast.literal_eval(text)
    except:
        return []


def convert_job_skills(text):

    if pd.isna(text):
        return []

    return [
        skill.strip()
        for skill in str(text).split(",")
        if skill.strip()
    ]


df["Resume Skills"] = df["Resume Skills"].apply(convert_resume_skills)

df["Required Skills"] = df["Required Skills"].apply(convert_job_skills)
```

```python
resources = {

    "python":"Python for Everybody (Coursera)",

    "java":"Java Programming (Udemy)",

    "sql":"SQLBolt",

    "aws":"AWS Skill Builder",

    "docker":"Docker Official Docs",

    "react":"React Official Docs",

    "mongodb":"MongoDB University",

    "tensorflow":"TensorFlow Tutorials",

    "pytorch":"PyTorch Tutorials",

    "git":"GitHub Learning Lab",

    "linux":"Linux Journey",

    "machine learning":"Andrew Ng ML Course",

    "deep learning":"Deep Learning Specialization",

    "fastapi":"FastAPI Documentation"

}
```

```python
results = []

for _, row in df.iterrows():

    resume = {x.lower().strip() for x in row["Resume Skills"]}

    required = {x.lower().strip() for x in row["Required Skills"]}

    matched = sorted(resume & required)

    missing = sorted(required - resume)

    if len(required) == 0:
        percentage = 0
    else:
        percentage = round(
            (len(matched)/len(required))*100,
            2
        )

    if percentage >= 80:
        strength = "Excellent"

    elif percentage >= 60:
        strength = "Good"

    elif percentage >= 40:
        strength = "Average"

    else:
        strength = "Needs Improvement"

    learning = []

    for skill in missing:

        learning.append(
            resources.get(
                skill,
                f"Learn {skill.title()} from Coursera/Udemy"
            )
        )

    if percentage >= 80:
        priority = "Low"

    elif percentage >= 60:
        priority = "Medium"

    else:
        priority = "High"

    results.append({

        "Resume": row["Resume"],

        "Job Title": row["Job Title"],

        "Company": row["Company"],

        "Similarity Score": row["Similarity Score"],

        "Skill Match %": percentage,

        "Resume Strength": strength,

        "Matched Skills": ", ".join(matched),

        "Missing Skills": ", ".join(missing),

        "Learning Recommendation": " | ".join(learning),

        "Learning Priority": priority

    })
```

```python
skill_gap = pd.DataFrame(results)

skill_gap.head()
```

```python
skill_gap = skill_gap.sort_values(
    by=["Skill Match %","Similarity Score"],
    ascending=False
)

skill_gap.head(20)
```

```python
skill_gap.to_csv(
    os.path.join(output_path,"skill_gap.csv"),
    index=False
)

print("Skill Gap Analysis Saved Successfully!")
```
