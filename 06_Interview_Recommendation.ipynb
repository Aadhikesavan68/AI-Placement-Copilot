```python
from google.colab import drive
drive.mount('/content/drive')

import pandas as pd
import random
```

```python
BASE="/content/drive/MyDrive/AI-Placement-Copilot-ML/datasets/processed"

company_df=pd.read_csv(BASE+"/company_questions.csv")

hr_df=pd.read_csv(BASE+"/hr_questions.csv")

print(company_df.shape)
print(hr_df.shape)
```

```python
recommended_company="tcs"

missing_skills=[
    "Arrays",
    "Core DSA"
]
```

```python
company_questions=company_df[
    company_df["company"].str.lower()==recommended_company.lower()
]

print(len(company_questions))
```

```python
filtered_questions=company_questions[
    company_questions["topic_tags"]
    .str.contains(
        "|".join(missing_skills),
        case=False,
        na=False
    )
]

print(len(filtered_questions))
```

```python
if len(filtered_questions)==0:
    filtered_questions=company_questions
```

```python
priority = {
    "High":3,
    "Medium":2,
    "Low":1
}

filtered_questions["priority_score"] = (
    filtered_questions["revision_priority"]
    .map(priority)
    .fillna(0)
)
```

```python
recommended = (
    filtered_questions
    .sort_values(
        by=["priority_score","frequency_%"],
        ascending=[False,False]
    )
    .head(10)
)

recommended
```

```python
recommended_hr = hr_df.sample(
    min(5,len(hr_df)),
    random_state=42
)

recommended_hr
```

```python
recommended.to_csv(
    BASE+"/recommended_company_questions.csv",
    index=False
)

recommended_hr.to_csv(
    BASE+"/recommended_hr_questions.csv",
    index=False
)
```
